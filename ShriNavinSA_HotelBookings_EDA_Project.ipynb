## 5. Handle Missing Values

```python
# Check missing values in every column
df.isnull().sum()
```

```python
# Check missing values as a percentage
(df.isnull().sum() / len(df)) * 100
```

```python
# Handle missing values

# 'company' has too many missing values so drop the column
if 'company' in df.columns:
    df = df.drop(columns=['company'])

# 'agent' -> missing means no agent was involved, fill with 0
if 'agent' in df.columns:
    df['agent'] = df['agent'].fillna(0)

# 'country' -> fill missing with the most common country (mode)
if 'country' in df.columns:
    df['country'] = df['country'].fillna(df['country'].mode()[0])

# 'children' -> very few missing values, fill with 0
if 'children' in df.columns:
    df['children'] = df['children'].fillna(0)
```

```python
# Confirm missing values have been handled
df.isnull().sum()
```

**Reason for choosing this method:**
- `company` had a very high percentage of missing values, so keeping it would add little value — it was dropped.
- `agent`, `children` had only a small number of missing values, and a missing value likely means "0" (no agent / no children), so filling with 0 made sense.
- `country` is categorical, so filling missing values with the most frequent value (mode) is a simple and reasonable beginner-friendly approach.


## 6. Clean the Dataset

```python
# Correct incorrect data types
# 'children', 'agent' should be integers, not floats
df['children'] = df['children'].astype(int)
df['agent'] = df['agent'].astype(int)

# Combine year, month, day into a proper date column
df['reservation_status_date'] = pd.to_datetime(df['reservation_status_date'])
```

```python
# Remove unwanted symbols / extra spaces from text columns
text_columns = df.select_dtypes(include='object').columns

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()   # remove extra spaces
    df[col] = df[col].str.lower()                # standardize case
```

```python
# Standardize different spellings of the same category (example: meal column)
df['meal'] = df['meal'].replace({'undefined': 'sc'})   # 'undefined' and 'sc' both mean no meal package
```

```python
# Handle invalid values
# adr (average daily rate) should not be negative
df = df[df['adr'] >= 0]

# Remove rows where total guests = 0 (no adults, children, or babies)
df = df[(df['adults'] + df['children'] + df['babies']) > 0]
```

## 9. Detect and Handle Outliers

```python
plt.figure(figsize=(6,4))
sns.boxplot(data=df, y='adr')
plt.title('Boxplot of ADR (Before Handling Outliers)')
plt.show()
```

```python
# Detect outliers using the IQR method
Q1 = df['adr'].quantile(0.25)
Q3 = df['adr'].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)

outlier_count = df[(df['adr'] < lower_limit) | (df['adr'] > upper_limit)].shape[0]
print("Number of outliers in ADR:", outlier_count)
```

```python
# Handle outliers: cap extreme values instead of removing rows
# (Capping is preferred here so we don't lose too much data)
df['adr'] = np.where(df['adr'] > upper_limit, upper_limit, df['adr'])
df['adr'] = np.where(df['adr'] < lower_limit, lower_limit, df['adr'])
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(data=df, y='adr')
plt.title('Boxplot of ADR (After Handling Outliers)')
plt.show()
```

**Explanation:** Outliers in `adr` were handled using the IQR (Interquartile Range) method. Instead of deleting these rows (which would lose valuable booking information), extreme values were **capped** at the lower/upper limits. This reduces the effect of extreme values while keeping the overall dataset size intact.
